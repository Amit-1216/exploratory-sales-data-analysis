### Data Exploration and Visualization

```python
# Summary of the data
sales.describe()
```

```python
#Q1. Print out, then visualize the most popular payment method.
sales['Payment Method'].value_counts() #.head(1)
```

```python
#Viz
base_color = sb.color_palette()[0]
method = sales['Payment Method'].value_counts().index
# method
sb.countplot(data = sales, y = 'Payment Method', color = base_color, order = method)
```

```python
#Q2. Print out, then visualize the most popular item purchased.
sales['Item Purchased'].value_counts()
```

```python
#Viz
item = sales['Item Purchased'].value_counts().index
# item
sb.countplot(data = sales, y = 'Item Purchased', color = base_color, order = item)
```

```python
# Print out the most popular item purchased by females
female_purchased = sales[sales['Gender'] == 'Female']
female_purchased['Item Purchased'].value_counts().head(1)
```

```python
#Q4. Visualize the locations that drive majority of the sales
location = sales['Location'].value_counts().head(10).index
sb.countplot(data = sales, y='Location', color = base_color, order=location)
```

```python
#Q5. On average how much do customers spend

print('Mean:',round(sales['Purchase Amount (USD)'].mean(),2))
print('Median:',round(sales['Purchase Amount (USD)'].median(),2))
```

```python
# Q6 – Which age group spends the most on shopping?

bins = [0, 18, 25, 35, 45, 60, 100]
labels = ['<18', '18-25', '26-35', '36-45', '46-60', '60+']

sales['Age Group'] = pd.cut(sales['Age'], bins=bins, labels=labels)

age_group_spending = sales.groupby('Age Group')['Purchase Amount (USD)'].mean()
age_group_spending
```

```python
# Q7 – Does gender influence average purchase amount?

gender_spending = sales.groupby('Gender')['Purchase Amount (USD)'].mean()
gender_spending
```

```python
# Q8 – What are the most purchased product categories?

category_counts = sales['Category'].value_counts()
category_counts
```

```python
# How does season affect shopping behavior?

season_sales = sales.groupby('Season')['Purchase Amount (USD)'].mean()
season_sales
```

```python
# Which locations generate the highest total sales?

location_sales = sales.groupby('Location')['Purchase Amount (USD)'].sum().sort_values(ascending=False)
location_sales.head(10)
```

```python
# What is the relationship between age and spending?

sales[['Age', 'Purchase Amount (USD)']].corr()
```

```python
# Does subscription status impact customer spending?

subscription_spending = sales.groupby('Subscription Status')['Purchase Amount (USD)'].mean()
subscription_spending
```

```python
# Do customers who use promo codes spend more?

promo_spending = sales.groupby('Promo Code Used')['Purchase Amount (USD)'].mean()
promo_spending
```

```python
# How does discount application affect spending?

discount_spending = sales.groupby('Discount Applied')['Purchase Amount (USD)'].mean()
discount_spending
```

```python
# What is the distribution of customer review ratings?

rating_distribution = sales['Review Rating'].value_counts().sort_index()
rating_distribution
```

```python
# Is there a relationship between review rating and spending?

rating_spending = sales.groupby('Review Rating')['Purchase Amount (USD)'].mean()
rating_spending
```

```python
# How does purchase frequency vary by age group?

age_purchase_frequency = sales.groupby('Age Group')['Previous Purchases'].mean()
age_purchase_frequency
```

```python
# Which shipping type is most commonly used?

shipping_counts = sales['Shipping Type'].value_counts()
shipping_counts
```

Key Insights & Conclusion:

- Middle-aged customers (26–45) contribute the highest average spending.
- Subscribed customers spend more compared to non-subscribers.
- Seasonal trends indicate higher purchases during winter and festive periods.
- Discounts and promo codes positively influence purchase value.

This analysis can help businesses optimize marketing strategies,
target high-value customers, and improve customer retention.
